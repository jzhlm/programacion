```python
# Definimos la lista original de números dada en el ejercicio
numeros = [15, 3, 8, 22, 7, 14, 9, 30, 5, 18]

# Creamos una lista vacía para guardar los resultados procesados
resultados = []

# Imprimimos un encabezado informativo para dar formato a la salida
print("=== INICIO DEL PROCESAMIENTO DE NÚMEROS ===\n")

# Iniciamos un bucle for para recorrer cada número de la lista
for num in numeros:
    # Verificamos si el número es menor a 10
    if num < 10:
        # Mostramos en pantalla el motivo por el cual se salta el número
        print(f"El número {num} fue saltado: es menor a 10.")
        # Usamos continue para pasar inmediatamente al siguiente elemento del ciclo
        continue

    # Verificamos si el número es impar usando el operador módulo (%)
    if num % 2 != 0:
        # Mostramos en pantalla el motivo por el cual se salta el número
        print(f"El número {num} fue saltado: es impar.")
        # Usamos continue para pasar al siguiente ciclo sin ejecutar lo de abajo
        continue

    # Si pasa ambos filtros, calculamos el cuadrado del número (num elevado a la 2)
    cuadrado = num ** 2

    # Agregamos el resultado obtenido a nuestra lista de resultados
    resultados.append(cuadrado)

    # Imprimimos el procesamiento exitoso mostrando el valor original y su cuadrado
    print(f"¡Éxito! El número {num} pasó los filtros. Su cuadrado es: {cuadrado}")

# Imprimimos una línea divisoria para organizar la presentación
print("\n" + "="*45)

# Mostramos la lista final con todos los cuadrados calculados
print(f"Resultados finales (cuadrados procesados): {resultados}")
```

```python
# Guardamos el texto dado en las instrucciones dentro de una variable multilínea
texto = """
Python es un lenguaje de programación poderoso y fácil de aprender.
Python tiene una sintaxis simple que permite escribir código claro.
Python es usado en ciencia de datos, web desarrollo, y automation.
"""

# Convertimos todo el texto a minúsculas para evitar diferencias por mayúsculas
texto_limpio = texto.lower()

# Reemplazamos los puntos y comas por texto vacío para limpiar la puntuación
texto_limpio = texto_limpio.replace(".", "").replace(",", "")

# Convertimos la cadena de texto en una lista de palabras individuales usando split()
palabras = texto_limpio.split()

# Obtenemos el total de palabras contando los elementos de la lista con len()
total_palabras = len(palabras)

# Convertimos la lista a un conjunto (set) para obtener únicamente las palabras sin duplicados
palabras_unicas = set(palabras)

# Obtenemos la cantidad de palabras únicas midiendo la longitud del conjunto
cantidad_unicas = len(palabras_unicas)

# Buscamos la palabra más larga utilizando max() con la función len como criterio de clave
palabra_mas_larga = max(palabras, key=len)

# Creamos un diccionario vacío para almacenar la frecuencia de cada palabra
frecuencia_palabras = {}

# Iniciamos un bucle for para contar la frecuencia de cada palabra en la lista
for palabra in palabras:
    # Si la palabra ya está en el diccionario, incrementamos su contador en 1
    if palabra in frecuencia_palabras:
        frecuencia_palabras[palabra] += 1
    # Si la palabra no existe en el diccionario, la inicializamos con el valor 1
    else:
        frecuencia_palabras[palabra] = 1

# Encontramos la palabra más frecuente buscando la clave con el valor máximo en el diccionario
palabra_mas_frecuente = max(frecuencia_palabras, key=frecuencia_palabras.get)

# Ordenamos las palabras por su frecuencia de mayor a menor utilizando sorted()
palabras_ordenadas = sorted(frecuencia_palabras.items(), key=lambda item: item[1], reverse=True)

# Tomamos las primeras 5 palabras de la lista ordenada para obtener el Top 5
top_5_palabras = palabras_ordenadas[:5]

# Imprimimos los resultados obtenidos con formato claro y estructurado
print("=== RESULTADOS DEL ANÁLISIS DE TEXTO ===")
print(f"1. Total de palabras: {total_palabras}")
print(f"2. Cantidad de palabras únicas: {cantidad_unicas}")
print(f"3. Palabra más larga: '{palabra_mas_larga}'")
print(f"4. Palabra más frecuente: '{palabra_mas_frecuente}' (se repite {frecuencia_palabras[palabra_mas_frecuente]} veces)")
print("\n--- TOP 5 PALABRAS MÁS USADAS ---")

# Recorremos el Top 5 e imprimimos el puesto, la palabra y su frecuencia
for posicion, (palabra, conteo) in enumerate(top_5_palabras, start=1):
    print(f" {posicion}. '{palabra}': {conteo} veces")
```
