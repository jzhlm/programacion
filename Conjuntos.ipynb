```python
# ==========================================
# PARTE I: Definición y Corrección de Conjuntos
# ==========================================

# a) Conjunto completo de 30 personas (Mexicanos)
personas = set([
    "Ana", "Luis", "María", "Carlos", "Elena", "Jorge", "Patricia", "Ricardo", "Sofía", "Fernando",
    "Gabriela", "Alejandro", "Claudia", "Roberto", "Isabel", "Manuel", "Teresa", "Héctor", "Julia", "Andrés",
    "Paula", "Miguel", "Verónica", "Arturo", "Lorena", "Esteban", "Carmen", "Raúl", "Adriana", "Diego"
])

# b) Corrección de conjuntos con los datos proporcionados:
# Nota: 'Marco' no pertenece al grupo inicial de 30 personas, pero se mantiene como parte del registro de Francia.
espana = set(["Ana", "Luis", "María", "Carlos", "Elena", "Jorge", "Patricia", "Ricardo", "Sofía", "Fernando"])

francia = set([
    "Ana", "Ricardo", "Claudia", "Sofía", "Isabel", "Manuel", "Teresa",
    "Héctor", "Julia", "Andrés", "Paula", "Miguel", "Marco"
])

# Se corrige la sintaxis de alemania cambiando la lista (...) a set(...)
alemania = set([
    "Carlos", "Elena", "Ricardo", "Isabel", "Manuel", "Teresa",
    "Héctor", "Ana", "Verónica", "Arturo", "Sofía", "Esteban"
])

# c) Imprimir los conjuntos principales
print("=== CONJUNTOS DE PERSONAS Y DESTINOS ===")
print("Personas (Total):", personas)
print("España:", espana)
print("Francia:", francia)
print("Alemania:", alemania)

# d) Funciones de Intersección entre combinaciones
E_inter_F = espana.intersection(francia)
E_inter_A = espana.intersection(alemania)
F_inter_A = francia.intersection(alemania)
E_inter_F_inter_A = espana.intersection(francia).intersection(alemania)

print("\n=== INTERSECCIONES ENTRE PAÍSES ===")
print("España ∩ Francia:", E_inter_F)
print("España ∩ Alemania:", E_inter_A)
print("Francia ∩ Alemania:", F_inter_A)
print("España ∩ Francia ∩ Alemania:", E_inter_F_inter_A)

# e) Longitud de las intersecciones
len_EF = len(E_inter_F)
len_EA = len(E_inter_A)
len_FA = len(F_inter_A)
len_EFA = len(E_inter_F_inter_A)

print("\n=== CANTIDAD DE PERSONAS POR INTERSECCIÓN ===")
print("Longitud (España ∩ Francia):", len_EF)
print("Longitud (España ∩ Alemania):", len_EA)
print("Longitud (Francia ∩ Alemania):", len_FA)
print("Longitud (España ∩ Francia ∩ Alemania):", len_EFA)


# ==========================================
# PARTE II: Diagrama de Venn
# ==========================================

# a) Instalación e importación de librerías
!pip install matplotlib-venn

import matplotlib.pyplot as plt
from matplotlib_venn import venn3

# b) Cálculo de regiones exclusivas (desglose de zonas para venn3)

# Triple intersección (España ∩ Francia ∩ Alemania)
solo_EFA = len_EFA  # 3 ("Ana", "Ricardo", "Sofía")

# Intersecciones dobles exclusivas (restando la triple intersección)
solo_EF = len_EF - solo_EFA  # 3 - 3 = 0
solo_EA = len_EA - solo_EFA  # 5 - 3 = 2
solo_FA = len_FA - solo_EFA  # 6 - 3 = 3

# Elementos que solo visitaron un país (restando intersecciones dobles y triples)
solo_E = len(espana) - (solo_EF + solo_EA + solo_EFA)  # 10 - (0 + 2 + 3) = 5
solo_F = len(francia) - (solo_EF + solo_FA + solo_EFA)  # 13 - (0 + 3 + 3) = 7
solo_A = len(alemania) - (solo_EA + solo_FA + solo_EFA) # 12 - (2 + 3 + 3) = 4

# c) Generar y mostrar el diagrama de Venn
plt.figure(figsize=(8, 8))
venn = venn3(
    subsets=(solo_E, solo_F, solo_EF, solo_A, solo_EA, solo_FA, solo_EFA),
    set_labels=("España", "Francia", "Alemania")
)
plt.title("Diagrama de Venn - Viajeros Mexicanos en Europa", fontsize=14)

# d) Guardar la imagen en tu sesión de Colab
plt.savefig("diagrama_de_venn_viajeros.png", dpi=300, bbox_inches='tight')
plt.show()
```
